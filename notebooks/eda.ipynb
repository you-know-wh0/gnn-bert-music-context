# FMA-Medium Preprocessing EDA

This notebook demonstrates how to load the preprocessed `.npz` files generated during the audio preprocessing step, inspect their structures and dimensions, and visualize their content.

```python
import numpy as np
import matplotlib.pyplot as plt

# Load a sample preprocessed npz file
npz_path = '../data/processed/audio_features/000002.npz'
data = np.load(npz_path)

print("Keys in file:", data.files)
print("Segmented Mel Spectrogram shape:", data['mel'].shape)
print("Segmented Chroma Feature shape:", data['chroma'].shape)
print("Sample rate:", data['sample_rate'])
print("Mel mean:", data['mel'].mean())
print("Chroma mean:", data['chroma'].mean())
```

## Visualizing Segmented Features

Let's plot the Mel Spectrogram and Chroma features for the first segment (5 seconds).

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 8))

# Plot Mel Spectrogram for Segment 0
im1 = axes[0].imshow(data['mel'][0], aspect='auto', origin='lower', cmap='viridis')
axes[0].set_title("Segment 0 - Log-Mel Spectrogram (128 bins)")
axes[0].set_ylabel("Mel Bins")
axes[0].set_xlabel("Time Frames")
fig.colorbar(im1, ax=axes[0])

# Plot Chroma for Segment 0
im2 = axes[1].imshow(data['chroma'][0], aspect='auto', origin='lower', cmap='coolwarm')
axes[1].set_title("Segment 0 - Chroma STFT (12 bins)")
axes[1].set_ylabel("Chroma Bins")
axes[1].set_xlabel("Time Frames")
fig.colorbar(im2, ax=axes[1])

plt.tight_layout()
plt.show()
```
